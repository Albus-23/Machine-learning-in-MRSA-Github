```python
# Lable data based on MIC values.
import numpy as np
import pandas as pd

df = pd.read_csv("MRSA_Project_venv/combined_data_cleaned.csv", index_col=None)


T = 5_000   # nM = 5 µM. Active if MIC <= T. This is your choice, so state it in your report.

v   = df["Standard Value"]
rel = df["Standard Relation"].str.strip("'\"")   # remove whitespace

active   = (v <= T) & rel.isin(["=", "<", "<="])
inactive = ((v > T) & (rel == "=")) | ((v >= T) & rel.isin([">", ">="]))

df["label"] = np.select([active, inactive], [1, 0], default=np.nan)
print(df["label"].value_counts(dropna=False))   # NaN = ambiguous rows
df = df.dropna(subset=["label"])
```

```python
df.to_csv("combined_labeled_data.csv", index=False)
```
